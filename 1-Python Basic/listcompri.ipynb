### list comprihension

basics of syantax   [expression for item in iterable]

with conditional logic [expression for item in iterable if condition]

nested list comprihension [expression for item1 in iterable1 for item2 in iterable2]

```python
#### basic list comprihension

square= [num ** 2 for num in range(10)]
print(square)
```

```python
lst = []
for i in range(10):
    if i % 2 == 0 :
        lst.append(i)
print(lst)
```

```python
even_numbers = [num for num in range(10) if num % 2 == 0 ]
print(even_numbers)
```

```python
lst1 = [1,2,3,4]
lst2 = ['a','b','c','d']

pair=[ (i,j) for i in lst1 for j in lst2 ]
print(pair)
```

```python
words = ["hello","world","python","list","comprehension"]
lengths = [len(word) for word in words]
print(lengths)
```
