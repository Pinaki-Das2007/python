```python
## while loop

count = 0
while count <= 5:
    print(count)
    count = count + 1
```

```python
count = 0 
while count % 2 == 0 :
    print(count)
    count = count + 1
```

```python
## Loop control statements
## break 
## The break statements exists the loop permanently

## break statement
for i in range(10):
    if i==5:
        break
    print(i)
```

```python
## continue 
## The continue statement skips the current itereation and continues with the next

for i in range(10):
    if i % 2 == 0 :
        continue
    print(i)
```

```python
## pass
## The pass state is a null statement , it does nothing.

for i in range(5):
    if i ==3:
        pass
    print(i)
```

```python
## nested loop
## a loop inside a loop
for i in range(3):
    for j in range(2):
        print(f"i:{i} and j:{j}")
```
