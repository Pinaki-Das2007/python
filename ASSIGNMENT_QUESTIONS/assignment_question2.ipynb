```python
num  = int (input("Enter the number of which you want to find the factorial:"))

factorial = 1
i = 1

while i <= num :
    factorial = factorial * i
    i += 1

print("The factorial of ", num , "is :" , factorial)
```

```python
num = int(input("Enter the number of 2 or mor digits:"))
rem = 0
sum = 0
while num > 0:
    rem = num % 10
    sum = rem + sum
    num = num // 10
print("The sum of the digits of the number is:", sum)
```

```python
num = int(input("Enter a number: "))

if num <= 1:
    print(num, "is not a prime number")
else:
    is_prime = True

    for i in range(2, num):
        if num % i == 0:
            is_prime = False
            break

    if is_prime:
        print(num, "is a prime number")
    else:
        print(num, "is not a prime number")
```

```python
n = int(input("Enter the number of terms: "))

a, b = 0, 1

print("Fibonacci Series:")

for i in range(n):
    print(a, end=" ")
    a, b = b, a + b
```
