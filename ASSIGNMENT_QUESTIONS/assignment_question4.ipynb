```python
dup_numbers = (1,3,2,2,4,5,6,6,7,8,1,9,10)
element = int(input("Enter the element to search: "))


count = dup_numbers.count(element)
print("Occurrences of",element,"=",count)


if element in dup_numbers:
    index = dup_numbers.index(element)
    print("First occurrence of",element,"is at index",index)
else:
    print(element,"not found in the tuple.")
```

3.15

```python
def min_in_tuple(tpl):
    return min(tpl)

def max_in_tuple(tpl):
    return max(tpl)

def sum_of_tuple(tpl):
    return sum(tpl)

sample_tpl = (1, 2, 3, 4, 5)
print(f"Minimum: {min_in_tuple(sample_tpl)}")
print(f"Maximum: {max_in_tuple(sample_tpl)}")
print(f"Sum: {sum_of_tuple(sample_tpl)}")
```
