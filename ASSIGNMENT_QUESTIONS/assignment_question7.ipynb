Assignment 1 : Fibonacci Sequence with Memoicatin


```python
def fibonacci(n, memo={}):
    if n in memo:
        return memo[n]
    if n <=1:
        return n
    memo[n] = fibonacci(n-1, memo) + fibonacci(n-2, memo)
    return memo[n]

print(fibonacci(10))
print(fibonacci(15))
```

Assignment 2 : Function with Nested Default Arguments

Define a function that takes two arguments: a and b, where b is a dictionary with a default value of an empty deictionary. The function should add a new key-value pair to the dictionary and return it. Test the function with different inputs.

```python
def add_to_dict(a, b=None):
    if b is None:
        b = {}
    b[a] = a**2
    return b

print(add_to_dict(2))
print(add_to_dict(3, {1: 1}))
```
