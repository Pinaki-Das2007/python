# Classes and Objects in Python

Object-oriented programming (OOP) is a programming paradigm that uses "objects" to desgin applications and computer programs. OOP allows for moderling real-world scenarios using classes and objects. This lesson covers the basics of creating classes and objects, including variables and methods.

```python
### A class is a blue print for creatin objects. Attributes, methods.

class car :
    pass

audi = car()
bmw = car()


print(type(audi))
```

```python
## Instance Variable and Methods
class dog:
    # constructor
    def __init__(self,name, age):
        self.name=name
        self.age = age

## Defining methods
dog1 = dog("Tommy",5)
print(dog1)
print(dog1.name)
print(dog1.age)

        

```

```python
dog2= dog("Lucy",2)
print(dog2)
print(dog2.name)
print(dog2.age)
```

```python
## Defina a class with instance methods

class dog:
    def __init__(self,name,age):
        self.name=name
        self.age=age

    def bark(self):
        print(f"{self.name} says woof")

dog1 = dog("Buddy",3)
dog1.bark()
dog2 = dog("Max",5)
dog2.bark()

```

```python
## Modeling  a Bank account

## Defining a class for bank account

class BankAccount:
    def __init__(self,owner,balance=0):
        self.owner=owner
        self.balance=balance

    def deposit(self,amount):
        self.balance += amount
        print(f"{amount} is deposited. New balance is {self.balance}")

    def withdraw(self,amount):
            if amount>self.balance: 
                print("Insufficient funds")
            else:
                self.balance = self.balance-amount
                print(f"{amount} is withdrawn. New balance is {self.balance}")


    def get_balance(self):
         return self.balance



## create an account

account = BankAccount("Pinaki",5000)
print(account.balance)
```

```python
## call instance variable

account.deposit(2000)
```

```python
account.withdraw(1000)
```

```python
print(account.get_balance())
```
