## SQL AND SQLITE

SQL (Structured Query Language) is a standard language for managing and manipulating relational databses. SQLite is a self-contained,serverless, and zero-congfiguration SQL database engine that is widely used for embedded database systems. In this lesson, we will covet the basics of SQL and SQLite, including how to creating databases, tables, and performing various SQL operations .

```python
import sqlite3
```

```python
## Connect to an SQLite database
connection = sqlite3.connect('example1.db')
connection
```

```python
cursor = connection.cursor()
```

```python
## Create a table
cursor.execute('''
Create Table If Not Exists employees(
id INTEGER PRIMARY KEY,
name TEXT NOT NULL,
age Integer,
department text )
''')

connection.commit()
```

```python
cursor.execute('''
Select * from employees
''')
```

```python
## Insert the data in sqlite table
cursor.execute('''
Insert into employees(name, age, department) 
values("pinaki", 23, "TL")
''')

cursor.execute('''
Insert into employees(name, age, department) 
values("Prakash", 24, "IT")
''')

cursor.execute('''
Insert into employees(name, age, department) 
values("Pritish", 22, "Developer")
''')

connection.commit()
cursor.execute('''
Select * from employees
''')
```

```python
## Quering the data from sqlite table
cursor.execute('Select * from employees')
rows=cursor.fetchall()

for row in rows:
    print(row)
```

```python
## Update the data in table
cursor.execute('''
UPDATE employees SET age=25 WHERE name="pinaki"
''')

connection.commit()
```

```python
cursor.execute('Select * from employees')
rows= cursor.fetchall()

for row in rows:
    print(row)
```
