```python
## important common dictionary methods
student = {"name": "krish", "age": 20, "grade": 'A'}

keys = student.keys()
print(keys)

values = student.values()
print(values)


items = student.items()
print(items)

```

```python
## shallow copy 

student_copy = student
print(student)
print(student_copy)


```

```python
student["name"] = "pinaki2"
print(student)
print(student_copy)
```

```python
student_copy1 = student.copy() ## shallow copy
print(student)
print(student_copy1)
```

```python
student["name "] = "pinaki3"
print(student)
print(student_copy1)
```

```python
## iterting through dictionary
## You can use loops to iterate through the keys, values, or key-value pairs of a dictionary.

for keys in student.keys():
    print(keys)
```

```python
## nested dictionary

students = {
    "student1":{"name":"pinaki","age":23,},
    'student2':{"name":"peter parker ","age":24}

}

print(students)
print(type(students))


```

```python
## Acess nested dictionary
print(students["student2"]["name"])
print(students["student2"]["age"])

```

```python
students.items()
```

```python
for student_id,student_info in students.items():
    print(f"{student_id}:{student_info}")
    for key,value in student_info.items():
        print(f"{key}:{value}")
```

```python
## dictionary comprehension

squares = {x:x**2 for x in range(5)}
print(squares)

```

```python
## conditional dictionary comprehension

evens = {x:x**2 for x in range(10) if x%2==0}
print(evens)
```

```python
## practical examples
## use a dictionary to count the frequency of elements in a list


numbers = [1,2,2,3,3,3,4,4,4,4]
frequency = {}

for number in numbers :
    if number in frequency:
        frequency [number ] += 1

    else :
        frequency [number ] =1
print(frequency)
```

```python
## merge 2 dictionaries into one

duct1 = {"a":1,"b":2}
duct2 = {"b":3,"c":4}

merged_dict = {**duct1,**duct2}
print(merged_dict)

```
