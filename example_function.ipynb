Example 1 : Temperature Conversion

```python
def temp(temp,unit):
    """This function is converting the temperature in to farenheit and celcius"""
    if unit == 'C': 
        return temp * 9/5 + 32    # Celcius to Farenheit
    elif unit == 'F':
        return temp -32 * 5/9      # Farenheit to Celcius
    else:
        return None

print(temp(25,'C'))
print(temp(77,'F'))

```

Example 2 : Password strength checker


```python
def pass_Strength(password):
    """ This function checks if the password is strong or not"""
    if len(password)<8:
        return False
    if not any(char.isdigit() for char in password ):
        return False
    if not any(char.isupper() for char in password):
        return False
    if not any(char.islower() for char in password):
        return False
    if not any(char in "!@#$%^&*()_+" for char in password):
        return False
    return True

## calling the function
print(pass_Strength("Weekpwd"))
print(pass_Strength("Weekpwd1"))
print(pass_Strength("Pinaki@2007"))

```

Example 3 : Check the total cost of items in a shopping cart


```python
def total_cost(cart):
    cost = 0
    for item in cart :
        cost += item['price']*item['quantity']
    return cost

## Example cart data
cart = [
    {'name':'Apple','price':0.5,'quantity':4},
    {'name':'Orange',"price":0.3,'quantity':6},
    {'name':'banana','price':0.4,'quantity':12}

]

## calling the function

cost = total_cost(cart)
print("The total payable cost is:",cost)
```

Example 4: Check if a string is a palindrome


```python
def is_palindrome(s):
    s = s.lower().replace(" "," ")
    return s==s[::-1] 

print(is_palindrome("A man, a plan, a canal: Panama"))
print(is_palindrome("hello"))
```

Example 5: Calculate the factorials of a number using recursion


```python
def factorial(n):
    if n==0:
        return 1
    else:
        return n*factorial(n-1)

## Calling the function
print(factorial(5))

```

Example 6: A function To read A file and count the frequency of each word in the file

```python
def count_word(file_path):
    word_count={}
    with open(file_path,'r') as file:
        for line in file:
            words = line.split()
            for word in words:
                word = word.lower().strip('.,!?:;()[]{}')
                word_count[word] = word_count.get(word,0)+1
        
    return word_count

    ## calling the functiom

file_path = 'sample.txt'
word_frequency = count_word(file_path)
print(word_frequency)
```

Example 7: Validate Email Address

```python
def valemail(email):
    if "@" in email and "." in email:
        return True
    else:
        return False

## calling the function
email = input("Enter the email address:")

if valemail(email):
    print("Valid email address")
else :
    print("Invalid email address")
```
