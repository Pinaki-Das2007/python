# File operations- Read And Write Files

File handling is a crucial part of any programming language. Python provides built-in functions and methods to rea from and write to files, both text and binary. This lesson will coveer the basics of file handling , including reading and writing text files and binary files.

```python
### Read a Whole file

with open('example.txt','r')as file :
    content=file.read()
    print(content)
```

```python
## Read a file line by line

with open('example.txt','r') as file:
    for line in file:
        print(line.strip())

## .strip() removes the newline character 
```

```python
## writing a file (Overwriting)

with open('example.txt','w') as file:
    file.write('Hello world!\n')
    file.write('My name is Pinaki\n')
    file.write('This is a new line')
```

```python
## Write a file without overwriting

with open('example.txt','a') as file:
    file.write("\nAppened operation taking place\n")
    file.write("\n Appened another line\n ")
```

```python
## Writing a list of lines to a file

lines = ['First line \n','second line \n','Third line \n']
with open('example.txt','a') as file:
    file.writelines(lines)
```

```python
### Binary files
# Writing to a binary file

data=b'\x00\x01\x02\x03\x04'
with open('example.bin','wb') as file:
    file.write(data)
    

```

```python
## Reading a binary file

with open('example.bin','rb') as file:
    content = file.read()
    print(content)
    
```

```python
### Read the content from a source text file and write it to a destination text file
# copying a text file

with open('example.txt','r') as source_file:
    content = source_file.read()

with open('destination.txt','w') as destination_file:
    destination_file.write(content)
```
