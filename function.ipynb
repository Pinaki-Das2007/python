```python
## why functions

num = 24
if num % 2 ==0:
    print("The number is even")
else:
    print("The number is odd")
```

```python
def even_or_odd(num):
    """This function finds even or odd"""
    if num % 2 == 0:
        print("The number is even ")
    else:
        print("The number is odd")
        
```

```python
## Call the function

even_or_odd(24)
```

```python
## Function with multiple parameters

def add(a,b):
    return a+b

result = add(2,3)
print(result)

```

```python
## default parameters

def greet(name):
    print(f"Hello {name} welcome to the paradise")

greet("Pinaki")
```

```python
def greet(name = "Guest"):
    print(f"Hello {name} welcome to the paradise")

greet("Krish")
```

```python
## Variable length arguments
## Postition and keywords arguments

def print_numbers(*krish):
    for numbers in krish:
        print(numbers)
```

```python
print_numbers(1,2,3,4,5,6,7,8,"krish")
```

```python
## keywords arguments

def print_details(**kwargs):
    
    for key, value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_details(name = "Krish", age = 23,country = "India")
```

```python

def print_details(*args,**kwargs):
    for val in args:
        print(f"Positional arguments:{val}")
    for key, value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_details(1,2,3,4,"krish",name="krish", age="32" , country="India")
```
