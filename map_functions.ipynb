The map() Function in Python


The map() function applies a given function to all items in an input list (or any other iterable) and returns a map objesct (an iterator). This is particularly useful for transforming data in a list comprehensively.

```python
def square(x):
    return x*x
    
square(10)
```

```python
numbers =[1,2,3,4,5,6,7,8]

list(map(square,numbers))
```

```python
## Lambda function with map

numbers=[1,2,3,4,5,6,7,8]
list(map(lambda x:x*x,numbers))
```
