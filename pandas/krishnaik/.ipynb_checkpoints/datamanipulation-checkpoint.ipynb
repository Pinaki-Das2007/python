# Data Manipulation and Analysis with Pandas

Data manipulation and analysis are key taska in data science or data analysis projects. Pandas provides a wide range of functions for data manipulation and analysis, making it easier to clean, transform, and extract insights from data. In this lesson, we will cover various data manipulation and analysis techniques using Pandas.

```python
import pandas as pd
```

```python
df = pd.read_csv('data.csv')
## fetch the first 5 rows of the dataframe
df.head(5)
```

```python
df_filled = df.fillna(0)
```

```python
## Filling missing values with mean of the column

df['Sales_filNa'] = df['Sales'].fillna(df['Sales'].mean())
df
```

```python
df.dtypes
```

```python
df = df.rename(columns={'Date':'Sales Date'})
df.head()
```

```python
df
```

```python
df['Value_new'] = df['Value'].fillna(df['Value'].mean()).astype(float)
df.head()
```

```python
df['New value'] = df['Value'].apply(lambda x:x * 2)
df.head()
```

```python
## Data agregating and grouping

grouped_mean = df.groupby('Product')['Sales'].mean()
print(grouped_mean)

```
