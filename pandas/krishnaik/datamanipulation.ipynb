```python
df_filled = df.fillna(0)
```

```python
## Filling missing values with mean of the column

df['Sales_filNa'] = df['Sales'].fillna(df['Sales'].mean())
df
```

```python
df.dtypes
```

```python
df = df.rename(columns={'Date':'Sales Date'})
df.head()
```

```python
df
```

```python
df['Value_new'] = df['Value'].fillna(df['Value'].mean()).astype(float)
df.head()
```

```python
df['New value'] = df['Value'].apply(lambda x:x * 2)
df.head()
```

```python
## Data agregating and grouping

grouped_mean = df.groupby('Product')['Sales'].mean()
print(grouped_mean)

```

```python
grouped_sum = df.groupby(['Product','Region'])['Value'].sum()
print(grouped_sum)
```

```python
grouped_mean = df.groupby(['Product','Region'])['Value'].mean()
print(grouped_mean)
```

```python
## Agreegate multiple functions

grouped_agg = df.groupby('Region')['Value'].agg(['mean','sum','count'])
print(grouped_agg)

```

```python
### Merging and joining DataFrames
# Create smaple DataFrames

df1 = pd.DataFrame({'Key':['A','B','C'], 'Value1':[1,2,3]})
df2 = pd.DataFrame({'Key':['A','B','D'], 'Value2':[4,5,6]})
```

```python
df1
```

```python
## Merge data frame on the key column

pd.merge(df1,df2,on='Key',how='inner')
```
