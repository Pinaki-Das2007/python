
# Pandas- DataFrame And Series

Pandas is a powerful data manipulation library in Python, widely used for data cleaning. It provides two primary data structures. DataFrame and structures: Series and dataFrame. A Series is a one-dimensional array-like object, while a DataFrame is a two-dimensional,size-mutable, and potentially hetrogeneous tabular data structure with labeled axes (rows and columns).

```python
import pandas as pd


```

```python
## Series
## Pandas Series is a one-dimensional array-like onject that can hold any data type. It is similar to a column in a table.

import pandas as pd
data=[1,2,3,4,5]
series = pd.Series(data)
print(series)
print(type(series))
print("Series \n",series)
```

```python
# Create a series from dictionary
data={"a":1,"b":2,"c":3}
series_dict =pd.Series(data) 
print(series_dict)
```

```python
data = [10,20,30]
index = ["a","b","c"]
pd.Series(data,index=index)
```

```python
## Dataframe
## Create a DataFrame from a dictionary of lists

data = {
    "Name":['Krish','John','Harry'],
    "Age":[25,30,45],
    "city":['Banglore','New York','Florida']
}

df = pd.DataFrame(data)
print(df)
print(type(df))
```
