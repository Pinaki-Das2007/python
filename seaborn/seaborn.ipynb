## Data Visualixation With Seaborn

Seaborn is a Python visualization library based on Matplotlib that provides a high-level interface for drawing attractive and informative statistical graphics. Seaborn helps in creating complex visualizations with just a few lines of code. In this lessonm we will cover the basics of Seaborn, including creating various types of plots and customizing them.

```python
import seaborn as sns
```

```python
## Basic Plotting with seaborn
tips =sns.load_dataset('tips')
tips
```

```python
## Create a scatter plot
import matplotlib.pyplot as plt

sns.scatterplot(x="total_bill",y="tip",data=tips)
plt.title("Scatter plot of total bill vs tips ")
plt.show()
```

```python
## Line plot
sns.lineplot(x="size",y="total_bill",data=tips)
plt.title("Line plot of total bill by size")
plt.show()
```

```python
## Categorical Plots
## Bar plot

sns.barplot(x="day",y="total_bill",data=tips)
plt.title("Bar Plot of total bill by day")
plt.show()
```

```python
## Box plot
sns.boxplot(x="day",y = "total_bill",data=tips)
plt.title("Box plot of total bill by day")
plt.show()
```

```python
## Violin Plot
sns.violinplot(x="day",y="total_bill",data=tips)
```

```python
### Histogram

sns.histplot(tips['total_bill'],bins=10,kde=True)
```

```python
## Kde plot
sns.kdeplot(tips['total_bill'],fill=True)
```

```python
## Pairplot
sns.pairplot(tips,)
```

```python
## Heatmap
corr = tips[['total_bill','tip','size']].corr()
corr
```

```python
sns.heatmap(corr,annot=True,cmap='coolwarm')
```
