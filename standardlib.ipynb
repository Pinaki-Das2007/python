# Standard Library Overview

 Python's Standard Library  is a vsst collectin of modules and packages that come bundled with Python, providing a wide range of functinlities out of the box. Here's an overview of some of the most commonly used modules and packages in the Python standard Library.


```python
import array
arr = array.array('i', [1, 2, 3, 4, 5])
print(arr)
```

```python
import math
print(math.sqrt(16))
print(math.pi)
```

```python
## random

import random
print(random.randint(1,10))
print(random.choice([1,2,3,4,5]))
```

```python
## File and directory access

import os
print(os.getcwd())
```

```python
os.mkdir("test_dir")
```

```python
## High level operations on files and collection files

import shutil
shutil.copyfile("source.txt","destination.txt")
```

```python
## Data serialization

import json
data = {'name':'pinaki','age':23}

json_str = json.dumps(data)
print(json_str)
print(type(json_str ))

parsed_data =json.loads(json_str)
print(parsed_data)
print(type(parsed_data))
```

```python
## csv

import csv

with open('example.csv', 'w', newline='') as file:
    writer = csv.writer(file)
    writer.writerow(['Name', 'Age'])
    writer.writerow(['Alice', 25])
    writer.writerow(['Bob', 30])

with open('example.csv', 'r') as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

```python
## date time

from datetime import datetime, timedelta

now = datetime.now()
print(now)

yesterday = now - timedelta(days=1)
print(yesterday)

tomorrow = now + timedelta(days=1)
print(tomorrow)
```

```python
## time

import time

print(time.time())

time.sleep(2)

print(time.time())
```

```python
## regular expression

import re

pattern = r'\d+'
text = "There are 123 apples"
match = re.search(pattern,text)
print(match.group())
```
