# Standard Library Overview

 Python's Standard Library  is a vsst collectin of modules and packages that come bundled with Python, providing a wide range of functinlities out of the box. Here's an overview of some of the most commonly used modules and packages in the Python standard Library.


```python
import array
arr = array.array('i', [1, 2, 3, 4, 5])
print(arr)
```

```python
import math
print(math.sqrt(16))
print(math.pi)
```

```python
## random

import random
print(random.randint(1,10))
print(random.choice([1,2,3,4,5]))
```

```python
## File and directory access

import os
print(os.getcwd())
```

```python
os.mkdir("test_dir")
```

```python
## High level operations on files and collection files

import shutil
shutil.copyfile("source.txt","destination.txt")
```

```python
## Data serialization

import json
data = {'name':'pinaki','age':23}

json_str = json.dumps(data)
print(json_str)
print(type(json_str ))
```
